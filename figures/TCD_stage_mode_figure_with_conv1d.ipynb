# TCD stage-supervision sweep + Conv1d reference

This notebook aggregates the three seeds for each `stage mode × Decoder RF`
condition and plots WBE and $P_{25}$ as mean $\pm$ one sample standard
deviation.

A separate Conv1d reference curve is loaded from
`conv1d_rf_sweep/confirm`. The Conv1d curve uses `conv_RF=13` so that the
scorer receptive field matches the stage-mode comparison.

By default, only points for which all three seeds are available are plotted.
Set `PLOT_PARTIAL_RESULTS = True` only for an interim preview while
experiments are still running.


```python
from itertools import product
from pathlib import Path
import re

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display


TCD_ROOT_DIR = Path(
    "/shared/data_zfs/blue2959/ND_Aligner/experiments/v2.2/TCD_stage_mode/confirm"
)

CONV1D_ROOT_DIR = Path(
    "/shared/data_zfs/blue2959/ND_Aligner/experiments/v2.2/conv1d_rf_sweep/confirm"
)

OUTPUT_PDF = TCD_ROOT_DIR / "tcd_stage_mode_with_conv1d.pdf"
OUTPUT_PNG = TCD_ROOT_DIR / "tcd_stage_mode_with_conv1d.png"
OUTPUT_SUMMARY_CSV = (
    TCD_ROOT_DIR / "tcd_stage_mode_with_conv1d_summary.csv"
)

SEEDS = [1234, 1235, 1236]
DECODER_RFS = [1, 5, 9, 13, 17]

# Conv1d reference에서 stage-mode experiment와 맞출 scorer RF
CONV1D_SCORER_RF = 13

# False: 세 seed가 모두 완료된 점만 사용 (논문용 기본값)
# True : 현재까지 완료된 seed만 평균하여 임시 확인
PLOT_PARTIAL_RESULTS = True


STEM_BASELINE_MODE = "first_only"
STEM_BASELINE_RF = 1

STAGE_MODE_ORDER = [
    # "geometric_w_st_0.5",
    "uniform",
    "geometric_w_st_2.0",
    "last_only",
]

STAGE_MODE_SPECS = {
    "first_only": {
        "label": "First-stage only",
        "color": "#0072B2",
        "linestyle": "-.",
        "marker": "P",
    },
    "geometric_w_st_0.5": {
        "label": r"$w_{\mathrm{st}}=0.5$",
        "color": "#56B4E9",
        "linestyle": "--",
        "marker": "^",
    },
    "uniform": {
        "label": r"Uniform ($w_{\mathrm{st}}=1.0$)",
        "color": "#009E73",
        "linestyle": "-",
        "marker": "o",
    },
    "geometric_w_st_2.0": {
        "label": r"$w_{\mathrm{st}}=2.0$",
        "color": "#E69F00",
        "linestyle": "--",
        "marker": "v",
    },
    "last_only": {
        "label": "Last-stage only",
        "color": "#D55E00",
        "linestyle": "-.",
        "marker": "X",
    },
}

CONV1D_STYLE = {
    "label": "Conv1d",
    "color": "#000000",
    "linestyle": "-",
    "marker": "s",
}


TCD_DIR_PATTERN = re.compile(
    r"^stage_mode=(?P<stage_mode>.+?)_"
    + r"dec_RF=(?P<decoder_rf>\d+)_"
    + r"seed=(?P<seed>\d+)$"
)

CONV1D_DIR_PATTERN = re.compile(
    r"^conv_RF=(?P<scorer_rf>[^_]+)_"
    + r"dec_RF=(?P<decoder_rf>\d+)_"
    + r"seed=(?P<seed>\d+)$"
)
```

```python
def load_tcd_stage_mode_results(
    root_dir: Path,
) -> pd.DataFrame:
    if not root_dir.exists():
        raise FileNotFoundError(
            f"실험 디렉터리가 없습니다: {root_dir}"
        )

    records = []

    for experiment_dir in sorted(root_dir.iterdir()):
        if not experiment_dir.is_dir():
            continue

        match = TCD_DIR_PATTERN.fullmatch(
            experiment_dir.name
        )

        if match is None:
            continue

        stage_mode = match.group("stage_mode")
        decoder_rf = int(match.group("decoder_rf"))
        seed = int(match.group("seed"))

        if stage_mode not in STAGE_MODE_SPECS:
            print(
                "[경고] 정의되지 않은 stage mode 제외:",
                experiment_dir.name,
            )
            continue

        if decoder_rf not in DECODER_RFS:
            continue

        if seed not in SEEDS:
            continue

        metric_path = experiment_dir / "eval_metrics.csv"

        if not metric_path.exists():
            print(
                "[대기] eval_metrics.csv 없음:",
                experiment_dir.name,
            )
            continue

        metric_df = pd.read_csv(metric_path)

        required_columns = {
            "word_boundary_error",
            "p_word_25ms",
        }
        missing_columns = (
            required_columns - set(metric_df.columns)
        )

        if missing_columns:
            raise ValueError(
                f"{metric_path}에 필요한 열이 없습니다: "
                f"{sorted(missing_columns)}"
            )

        if metric_df.empty:
            print(f"[경고] 비어 있는 CSV: {metric_path}")
            continue

        if len(metric_df) > 1:
            print(
                f"[경고] {experiment_dir.name}: "
                f"row가 {len(metric_df)}개이므로 "
                "마지막 row를 사용합니다."
            )

        row = metric_df.iloc[-1]

        records.append(
            {
                "stage_mode": stage_mode,
                "decoder_rf": decoder_rf,
                "seed": seed,
                "wbe_ms": (
                    float(row["word_boundary_error"])
                    * 1000.0
                ),
                "p25": float(row["p_word_25ms"]),
            }
        )

    return pd.DataFrame(records)


def load_conv1d_results(
    root_dir: Path,
    scorer_rf: int,
) -> pd.DataFrame:
    if not root_dir.exists():
        raise FileNotFoundError(
            f"Conv1d 실험 디렉터리가 없습니다: {root_dir}"
        )

    records = []

    for experiment_dir in sorted(root_dir.iterdir()):
        if not experiment_dir.is_dir():
            continue

        match = CONV1D_DIR_PATTERN.fullmatch(
            experiment_dir.name
        )

        if match is None:
            continue

        current_scorer_rf = match.group("scorer_rf")

        # l2 등의 다른 scorer setting은 제외하고,
        # stage-mode figure와 맞는 conv_RF만 사용
        if current_scorer_rf != str(scorer_rf):
            continue

        decoder_rf = int(match.group("decoder_rf"))
        seed = int(match.group("seed"))

        if decoder_rf not in DECODER_RFS:
            continue

        if seed not in SEEDS:
            continue

        metric_path = experiment_dir / "eval_metrics.csv"

        if not metric_path.exists():
            print(
                "[대기] Conv1d eval_metrics.csv 없음:",
                experiment_dir.name,
            )
            continue

        metric_df = pd.read_csv(metric_path)

        required_columns = {
            "word_boundary_error",
            "p_word_25ms",
        }
        missing_columns = (
            required_columns - set(metric_df.columns)
        )

        if missing_columns:
            raise ValueError(
                f"{metric_path}에 필요한 열이 없습니다: "
                f"{sorted(missing_columns)}"
            )

        if metric_df.empty:
            print(f"[경고] 비어 있는 CSV: {metric_path}")
            continue

        if len(metric_df) > 1:
            print(
                f"[경고] {experiment_dir.name}: "
                f"row가 {len(metric_df)}개이므로 "
                "마지막 row를 사용합니다."
            )

        row = metric_df.iloc[-1]

        records.append(
            {
                "decoder_rf": decoder_rf,
                "seed": seed,
                "wbe_ms": (
                    float(row["word_boundary_error"])
                    * 1000.0
                ),
                "p25": float(row["p_word_25ms"]),
            }
        )

    return pd.DataFrame(records)
```

```python
result_df = load_tcd_stage_mode_results(
    TCD_ROOT_DIR
)

if result_df.empty:
    raise RuntimeError(
        f"완료된 TCD eval_metrics.csv를 찾지 못했습니다: {TCD_ROOT_DIR}"
    )

result_df = (
    result_df
    .sort_values(
        ["stage_mode", "decoder_rf", "seed"]
    )
    .reset_index(drop=True)
)


conv1d_result_df = load_conv1d_results(
    CONV1D_ROOT_DIR,
    scorer_rf=CONV1D_SCORER_RF,
)

if conv1d_result_df.empty:
    raise RuntimeError(
        "완료된 Conv1d 결과를 찾지 못했습니다: "
        f"{CONV1D_ROOT_DIR} "
        f"(conv_RF={CONV1D_SCORER_RF})"
    )

conv1d_result_df = (
    conv1d_result_df
    .sort_values(["decoder_rf", "seed"])
    .reset_index(drop=True)
)


print("TCD stage-mode results")
display(result_df)
print(f"불러온 TCD 결과 수: {len(result_df)}")

print(
    f"\nConv1d reference results "
    f"(conv_RF={CONV1D_SCORER_RF})"
)
display(conv1d_result_df)
print(f"불러온 Conv1d 결과 수: {len(conv1d_result_df)}")
```

```python
# first_only에서는 RF=1 결과만 stem baseline으로 사용
analysis_result_df = result_df[
    (
        result_df["stage_mode"]
        != STEM_BASELINE_MODE
    )
    | (
        result_df["decoder_rf"]
        == STEM_BASELINE_RF
    )
].copy()


summary_df = (
    analysis_result_df
    .groupby(
        ["stage_mode", "decoder_rf"],
        as_index=False,
    )
    .agg(
        seed_count=("seed", "nunique"),
        wbe_mean=("wbe_ms", "mean"),
        wbe_std=("wbe_ms", "std"),
        p25_mean=("p25", "mean"),
        p25_std=("p25", "std"),
    )
)


ALL_MODE_ORDER = [
    STEM_BASELINE_MODE,
    *STAGE_MODE_ORDER,
]

summary_df["stage_mode"] = pd.Categorical(
    summary_df["stage_mode"],
    categories=ALL_MODE_ORDER,
    ordered=True,
)

summary_df = (
    summary_df
    .sort_values(["stage_mode", "decoder_rf"])
    .reset_index(drop=True)
)


# Stem-only baseline: first_only, Decoder RF=1
stem_baseline_df = summary_df[
    (
        summary_df["stage_mode"]
        == STEM_BASELINE_MODE
    )
    & (
        summary_df["decoder_rf"]
        == STEM_BASELINE_RF
    )
]

if len(stem_baseline_df) != 1:
    raise ValueError(
        "Stem-only baseline 결과를 정확히 하나 찾지 못했습니다."
    )

stem_baseline = stem_baseline_df.iloc[0]

if stem_baseline["seed_count"] != len(SEEDS):
    raise ValueError(
        "Stem-only baseline에 세 seed가 모두 필요합니다."
    )


curve_summary_df = summary_df[
    summary_df["stage_mode"].isin(
        STAGE_MODE_ORDER
    )
].copy()

if PLOT_PARTIAL_RESULTS:
    plot_summary_df = curve_summary_df.copy()
else:
    plot_summary_df = curve_summary_df[
        curve_summary_df["seed_count"]
        == len(SEEDS)
    ].copy()


# ------------------------------------------------------------
# Conv1d reference summary
# ------------------------------------------------------------
conv1d_summary_df = (
    conv1d_result_df
    .groupby(
        "decoder_rf",
        as_index=False,
    )
    .agg(
        seed_count=("seed", "nunique"),
        wbe_mean=("wbe_ms", "mean"),
        wbe_std=("wbe_ms", "std"),
        p25_mean=("p25", "mean"),
        p25_std=("p25", "std"),
    )
    .sort_values("decoder_rf")
    .reset_index(drop=True)
)

if PLOT_PARTIAL_RESULTS:
    conv1d_plot_df = conv1d_summary_df.copy()
else:
    conv1d_plot_df = conv1d_summary_df[
        conv1d_summary_df["seed_count"]
        == len(SEEDS)
    ].copy()


if plot_summary_df.empty:
    raise RuntimeError(
        "현재 TCD 그래프에 사용할 완료된 설정이 없습니다."
    )

if conv1d_plot_df.empty:
    raise RuntimeError(
        "현재 Conv1d 그래프에 사용할 완료된 설정이 없습니다."
    )


print("TCD stage-mode summary")
display(
    summary_df.round(
        {
            "wbe_mean": 3,
            "wbe_std": 3,
            "p25_mean": 3,
            "p25_std": 3,
        }
    )
)

print(
    "Stem-only baseline:"
    + f" WBE={stem_baseline['wbe_mean']:.3f}"
    + f" ± {stem_baseline['wbe_std']:.3f} ms,"
    + f" P25={stem_baseline['p25_mean']:.3f}"
    + f" ± {stem_baseline['p25_std']:.3f}%"
)

print(
    f"\nConv1d summary (conv_RF={CONV1D_SCORER_RF})"
)
display(
    conv1d_summary_df.round(
        {
            "wbe_mean": 3,
            "wbe_std": 3,
            "p25_mean": 3,
            "p25_std": 3,
        }
    )
)


# 하나의 CSV로 저장
tcd_csv_df = summary_df.copy()
tcd_csv_df.insert(0, "model", "TCD")

conv1d_csv_df = conv1d_summary_df.copy()
conv1d_csv_df.insert(0, "model", "Conv1d")
conv1d_csv_df.insert(
    1,
    "stage_mode",
    f"conv_RF={CONV1D_SCORER_RF}",
)

combined_summary_df = pd.concat(
    [tcd_csv_df, conv1d_csv_df],
    ignore_index=True,
)

combined_summary_df.to_csv(
    OUTPUT_SUMMARY_CSV,
    index=False,
)
```

```python
AXIS_LABEL_SIZE = 14
TICK_LABEL_SIZE = 12
LEGEND_SIZE = 10


fig, axes = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(10.8, 2.8),
    sharex=True,
)

metric_specs = [
    {
        "ax": axes[0],
        "mean_column": "wbe_mean",
        "std_column": "wbe_std",
        "ylabel": "WBE (ms) ↓",
    },
    {
        "ax": axes[1],
        "mean_column": "p25_mean",
        "std_column": "p25_std",
        "ylabel": r"$P_{25}$ (%) ↑",
    },
]

# ------------------------------------------------------------
# Stem-only horizontal baseline
# ------------------------------------------------------------

for spec in metric_specs:
    ax = spec["ax"]

    baseline_mean = float(
        stem_baseline[spec["mean_column"]]
    )
    baseline_std = float(
        stem_baseline[spec["std_column"]]
    )

    unit = (
        "ms"
        if spec["mean_column"] == "wbe_mean"
        else "%"
    )

    ax.axhline(
        y=baseline_mean,
        color="#666666",
        linestyle="--",
        linewidth=1.6,
        zorder=2,
    )

    is_p25 = (
        spec["mean_column"] == "p25_mean"
    )

    text_offset_y = -5 if is_p25 else 4
    text_va = "top" if is_p25 else "bottom"

    ax.annotate(
        (
            f"Stem-only: "
            f"{baseline_mean:.2f} "
            rf"$\pm$ {baseline_std:.2f} {unit}"
        ),
        xy=(0.98, baseline_mean),
        xycoords=(
            "axes fraction",
            "data",
        ),
        xytext=(0, text_offset_y),
        textcoords="offset points",
        ha="right",
        va=text_va,
        color="#555555",
        fontsize=10.5,
        bbox={
            "facecolor": "white",
            "edgecolor": "none",
            "alpha": 0.85,
            "pad": 1.5,
        },
        zorder=5,
    )


# ------------------------------------------------------------
# TCD stage-mode curves
# ------------------------------------------------------------

for stage_mode in STAGE_MODE_ORDER:
    mode_df = (
        plot_summary_df[
            plot_summary_df["stage_mode"] == stage_mode
        ]
        .set_index("decoder_rf")
        .reindex(DECODER_RFS)
    )

    if mode_df["wbe_mean"].isna().all():
        continue

    style = STAGE_MODE_SPECS[stage_mode]

    for spec in metric_specs:
        yerr = (
            mode_df[spec["std_column"]]
            .fillna(0.0)
        )

        spec["ax"].errorbar(
            x=mode_df.index,
            y=mode_df[spec["mean_column"]],
            yerr=yerr,
            label=style["label"],
            color=style["color"],
            linestyle=style["linestyle"],
            marker=style["marker"],
            markersize=6.0,
            linewidth=1.9,
            elinewidth=1.1,
            capsize=3.0,
            capthick=1.1,
            zorder=3,
        )


# ------------------------------------------------------------
# Conv1d reference curve
# ------------------------------------------------------------

conv1d_mode_df = (
    conv1d_plot_df
    .set_index("decoder_rf")
    .reindex(DECODER_RFS)
)

for spec in metric_specs:
    yerr = (
        conv1d_mode_df[spec["std_column"]]
        .fillna(0.0)
    )

    spec["ax"].errorbar(
        x=conv1d_mode_df.index,
        y=conv1d_mode_df[spec["mean_column"]],
        yerr=yerr,
        label=CONV1D_STYLE["label"],
        color=CONV1D_STYLE["color"],
        linestyle=CONV1D_STYLE["linestyle"],
        marker=CONV1D_STYLE["marker"],
        markersize=5.8,
        linewidth=2.0,
        elinewidth=1.1,
        capsize=3.0,
        capthick=1.1,
        zorder=4,
    )


for spec in metric_specs:
    ax = spec["ax"]

    ax.set_xlabel(
        "Decoder RF",
        fontsize=AXIS_LABEL_SIZE,
    )
    ax.set_ylabel(
        spec["ylabel"],
        fontsize=AXIS_LABEL_SIZE,
    )
    ax.set_xticks(DECODER_RFS)

    ax.tick_params(
        axis="both",
        which="major",
        labelsize=TICK_LABEL_SIZE,
        color="black",
        width=0.9,
    )

    ax.grid(
        axis="both",
        linestyle="--",
        linewidth=0.6,
        alpha=0.25,
        zorder=0,
    )

    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_color("black")
        spine.set_linewidth(0.9)


handles, labels = axes[0].get_legend_handles_labels()

legend = fig.legend(
    handles,
    labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 1.08),
    ncol=4,
    frameon=True,
    fancybox=True,
    framealpha=1.0,
    facecolor="white",
    edgecolor="#B0B0B0",
    fontsize=LEGEND_SIZE,
    columnspacing=1.15,
    handlelength=2.2,
    borderpad=0.6,
    labelspacing=0.4,
)

legend.get_frame().set_linewidth(0.8)

fig.tight_layout(
    rect=(0, 0, 1, 0.95)
)

fig.savefig(
    OUTPUT_PDF,
    bbox_inches="tight",
)

fig.savefig(
    OUTPUT_PNG,
    dpi=300,
    bbox_inches="tight",
)

plt.show()

print(f"PDF 저장 완료: {OUTPUT_PDF}")
print(f"PNG 저장 완료: {OUTPUT_PNG}")
```
