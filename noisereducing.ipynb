```python
import soundfile as sf
import noisereduce as nr

from tts.audio.utils.suppress_impulsive_peaks import suppress_impulsive_peaks
import torch

wav_path = "/shared/data_zfs/blue2959/Buckeye-grid/s05/s0502b_chunk_0028.wav"
out_path = "input_denoised.wav"

wav, sr = sf.read(wav_path, dtype="float32")

wav_denoised = nr.reduce_noise(
    y=wav,
    sr=sr,
    stationary=False,   # 시간에 따라 변하는 noise
    prop_decrease=0.8,  # 1.0은 너무 세게 깎일 수 있음
    n_fft=512,          # speech용
)
wav_denoised = suppress_impulsive_peaks(torch.tensor(wav_denoised)).detach().cpu().numpy()

wav_denoised = wav_denoised / wav_denoised.max() * 0.6


sf.write(out_path, wav_denoised, sr)
```
